````python
# -------------------------------------------------------------
# 📊 Visualize Distribution for 'children'
# -------------------------------------------------------------


# Calculate missing values
missing_children = df['children'].isnull().sum()
total_records = len(df)
missing_children_percent = (missing_children / total_records) * 100

# Plot histogram + density curve for the column
sns.histplot(df['children'], kde=True, color="#009FE3", bins=30, edgecolor='white')

# Chart styling
plt.title("Distribution of 'children' (Before Filling Missing Values)",
          fontsize=9, fontweight='bold', backgroundcolor="#E6F2FF", pad=15)
plt.xlabel("Number of Children", fontsize=12, color="#003580")
plt.ylabel("Frequency", fontsize=12, color="#003580")
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Add outer border for neat presentation
for spine in plt.gca().spines.values():
    spine.set_edgecolor("#003580")
    spine.set_linewidth(1.2)

# Add missing percentage note below chart
plt.figtext(0.5, -0.05,
            f"Missing Values in 'children': {missing_children:,} ({missing_children_percent:.2f}%) of total records",
            wrap=True, ha='center', fontsize=9,
            color="#003580", fontweight='bold', backgroundcolor="#E6F2FF")


plt.tight_layout()
plt.show()

````

**Data Type Correction**

````python
# -------------------------------------------------------------
# Convert Split Date Columns into a Single Datetime Column
# -------------------------------------------------------------

# Check if all three columns exist before creating the combined date
if {'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month'}.issubset(df_clean.columns):

    # Combine year, month, and day into one string column
    df_clean['arrival_date'] = (
        df_clean['arrival_date_year'].astype(str) + '-' +
        df_clean['arrival_date_month'].astype(str) + '-' +
        df_clean['arrival_date_day_of_month'].astype(str)
    )

    # Convert the combined string to a proper datetime object
    df_clean['arrival_date'] = pd.to_datetime(df_clean['arrival_date'], errors='coerce', format='%Y-%B-%d')

    print("arrival_date' column created successfully as datetime.")

else:
    print("Required arrival date columns not found — cannot create arrival_date.")

# -------------------------------------------------------------
# Convert 'reservation_status_date' directly to datetime
# -------------------------------------------------------------
if 'reservation_status_date' in df_clean.columns:
    df_clean['reservation_status_date'] = pd.to_datetime(df_clean['reservation_status_date'], errors='coerce')
    print("reservation_status_date' converted to datetime successfully.")
````

````python
# -------------------------------------------------------------
# Convert Categorical Columns to 'category' Data Type
# -------------------------------------------------------------

categorical_cols = [
    'hotel', 'meal', 'country', 'market_segment', 'distribution_channel',
    'reserved_room_type', 'assigned_room_type', 'deposit_type',
    'customer_type', 'reservation_status'
]

converted_cols = []  # To track which columns were converted

for col in categorical_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].astype('category')
        converted_cols.append(col)  # Log converted columns

# Print summary message
if converted_cols:
    print(f"Successfully converted {len(converted_cols)} columns to 'category' dtype:")
    print(" ", ', '.join(converted_cols))
else:
    print(" No categorical columns were found for conversion.")
````

````python
# -------------------------------------------------------------
#  Convert Numeric-Like Columns to Numeric (int/float)
# -------------------------------------------------------------

numeric_cols = ['adults', 'children', 'babies', 'agent', 'company', 'adr']

converted_numeric_cols = []   # Track successfully converted columns
not_found_cols = []           # Track columns missing in dataframe

for col in numeric_cols:
    if col in df_clean.columns:
        # Convert to numeric (float first to handle NaN safely)
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

        # If column is ID-like (agent/company), convert to int (while keeping NaN as is)
        if col in ['agent', 'company']:
            df_clean[col] = df_clean[col].astype('Int64')  # nullable integer

        converted_numeric_cols.append(col)
    else:
        not_found_cols.append(col)

# Print status output
if converted_numeric_cols:
    print(f"Successfully converted {len(converted_numeric_cols)} columns to numeric dtype:")
    print("   ➜", ", ".join(converted_numeric_cols))
else:
    print(" No numeric-like columns found to convert.")

#  Print missing column warnings (if any)
if not_found_cols:
    print(f"The following columns were not found in the dataframe and could not be converted:")
    print("   ➜", ", ".join(not_found_cols))
````

## **Data Transformation**

````python

# Create total guests column
if {'adults', 'children', 'babies'}.issubset(df_clean.columns):
    df_clean['total_guests'] = (
        df_clean['adults'] + df_clean['children'] + df_clean['babies']
    )

# Create total stay duration column
if {'stays_in_weekend_nights', 'stays_in_week_nights'}.issubset(df_clean.columns):
    df_clean['total_stay_nights'] = (
        df_clean['stays_in_weekend_nights'] + df_clean['stays_in_week_nights']
    )

# Extract year & month from 'reservation_status_date'
if 'reservation_status_date' in df_clean.columns:
    df_clean['reservation_year'] = df_clean['reservation_status_date'].dt.year
    df_clean['reservation_month'] = df_clean['reservation_status_date'].dt.month_name()

# Create categorical column for Corporate vs Individual Booking
if 'company' in df_clean.columns:
    df_clean['booking_type'] = df_clean['company'].apply(
        lambda x: 'Corporate Booking' if pd.notna(x) else 'Individual Booking'
    )

# Create categorical column for Direct vs Agent Booking
if 'agent' in df_clean.columns:
    df_clean['booking_channel'] = df_clean['agent'].apply(
        lambda x: 'Via Agent' if pd.notna(x) else 'Direct Booking'
    )

# Create total revenue column (ADR × total nights)
if {'adr', 'stays_in_week_nights', 'stays_in_weekend_nights'}.issubset(df_clean.columns):
    df_clean['total_revenue'] = (
        df_clean['adr'] *
        (df_clean['stays_in_week_nights'] + df_clean['stays_in_weekend_nights'])
    )

# Extract additional date-based insights
if 'arrival_date' in df_clean.columns:
    df_clean['arrival_month'] = df_clean['arrival_date'].dt.month_name()
    df_clean['arrival_day'] = df_clean['arrival_date'].dt.day_name()

# Print summary of created features
print(" Feature Engineering Completed Successfully!")
````

# **Detect & Handle Outliers Using Z-Score**

````python


# Create a copy to preserve original data
df_zscore = df_clean.copy()

# Define Z-score threshold
z_threshold = 3
outlier_summary = {}

# Compute Z-scores and count outliers
for col in numeric_cols:
    z_scores = np.abs(stats.zscore(df_zscore[col], nan_policy='omit'))
    outlier_count = np.sum(z_scores > z_threshold)
    outlier_summary[col] = outlier_count

    # Cap outliers with mean value
    df_zscore.loc[z_scores > z_threshold, col] = df_zscore[col].mean()

# =============================================================
# Display formatted summary table of detected outliers
# =============================================================

print("="*70)
print(f"{'OUTLIER DETECTION SUMMARY (Z-SCORE METHOD)':^70}")
print("="*70)
print(f"{'Feature':<35} | {'Outliers Detected':>20}")
print("-"*70)

total_outliers = 0
for feature, count in outlier_summary.items():
    print(f"{feature:<35} | {count:>20,}")
    total_outliers += count

print("-"*70)
print(f"{'Total Outliers Detected Across All Features':<35} | {total_outliers:>20,}")
print("="*70)
print("Outlier treatment successfully completed using Z-Score method.\n")
````

### **KPI Dashboard Prep**

````python
import calendar
import matplotlib.patches as patches

# --- Prepare Data ---
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'], errors='coerce')
df['month'] = df['reservation_status_date'].dt.month_name()
df['month_num'] = df['reservation_status_date'].dt.month
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['total_revenue'] = df['adr'] * df['total_nights']

# --- Compute KPIs ---
total_bookings = len(df)
cancel_rate = round(df['is_canceled'].mean() * 100, 1)
avg_adr = round(df['adr'].mean(), 2)
avg_stay = round(df['total_nights'].mean(), 1)
avg_lead = round(df['lead_time'].mean(), 1)
total_revenue = round(df['total_revenue'].sum() / 1e6, 2)

# --- Theme setup ---
sns.set_style("whitegrid")
blue_palette = ["#003580", "#005EB8", "#009FE3", "#33CCFF", "#99E6FF"]
background_color = "#F9FBFF"

fig = plt.figure(figsize=(15, 10))
fig.patch.set_facecolor(background_color)

# =============================================================
# (A) KPI Cards (Centered + Proper Spacing)
# =============================================================
ax0 = plt.subplot2grid((3, 4), (0, 0), colspan=4)
ax0.axis("off")

# Centered Title
ax0.set_title(
    "Booking.com — KPI Summary",
    fontsize=15,
    fontweight="bold",
    color="#003580",
    loc="center",
    pad=25
)

# KPI Data
kpi_data = {
    "Metric": ["Total Bookings", "Total Revenue (M$)", "Avg ADR ($)",
               "Cancellation Rate (%)", "Avg Stay (nights)", "Avg Lead Time (days)"],
    "Value": [f"{total_bookings:,}", total_revenue, avg_adr, cancel_rate, avg_stay, avg_lead]
}

# --- Layout Parameters ---
num_cards = len(kpi_data["Metric"])
card_width = 0.09
card_height = 0.42
gap = 0.05  # increased spacing between cards for better breathing room

total_width = num_cards * card_width + (num_cards - 1) * gap
start_x = 0.5 - total_width / 2  # perfectly centered

# --- Draw KPI Cards ---
for i, (metric, val) in enumerate(zip(kpi_data["Metric"], kpi_data["Value"])):
    x_pos = start_x + i * (card_width + gap)

    # Card (white background with blue border)
    card = patches.FancyBboxPatch(
        (x_pos, 0.35), card_width, card_height,
        boxstyle="round,pad=0.02,rounding_size=0.06",
        linewidth=1.8, edgecolor="#003580", facecolor="white",
        transform=ax0.transAxes, zorder=2
    )
    ax0.add_patch(card)

    # KPI Value
    ax0.text(x_pos + card_width / 2, 0.58, f"{val}",
             fontsize=13, fontweight="bold", color="#009FE3",
             ha="center", transform=ax0.transAxes, zorder=3)

    # KPI Label
    ax0.text(x_pos + card_width / 2, 0.42, metric,
             fontsize=9, color="#003580", ha="center", transform=ax0.transAxes, zorder=3)

# =============================================================
# (B) Average ADR by Month
# =============================================================
ax1 = plt.subplot2grid((3, 4), (1, 0), colspan=2)
adr_month = df.groupby('month_num')['adr'].mean().reindex(range(1, 13))
sns.lineplot(x=list(calendar.month_abbr)[1:], y=adr_month.values, ax=ax1,
             marker="o", color="#009FE3", linewidth=2.5)
ax1.set_title("Average ADR by Month", fontsize=11, fontweight="bold", color="#003580")
ax1.set_xlabel("")
ax1.set_ylabel("ADR ($)")
ax1.grid(alpha=0.3)

# =============================================================
# (C) Cancellation Rate by Hotel Type
# =============================================================
ax2 = plt.subplot2grid((3, 4), (1, 2), colspan=2)
cancel_rate_by_hotel = df.groupby('hotel')['is_canceled'].mean().sort_values(ascending=False) * 100
sns.barplot(x=cancel_rate_by_hotel.values, y=cancel_rate_by_hotel.index, palette=blue_palette, ax=ax2)
ax2.set_title("Cancellation Rate by Hotel Type", fontsize=11, fontweight="bold", color="#003580")
ax2.set_xlabel("Cancel Rate (%)")
ax2.set_ylabel("")
ax2.grid(axis='x', linestyle='--', alpha=0.3)

# =============================================================
# (D) Revenue Heatmap (Month × Hotel)
# =============================================================
# =============================================================
# (D) Revenue Heatmap (Month × Hotel) — With Month Names
# =============================================================
ax3 = plt.subplot2grid((3, 4), (2, 0), colspan=2)

# Prepare heatmap data
heat_data = df.groupby(['month_num', 'hotel'])['total_revenue'].mean().unstack()

# Plot heatmap
sns.heatmap(heat_data, cmap="YlGnBu", ax=ax3, annot=True, fmt=".0f", cbar=False)

# --- Custom month names for Y-axis ---
month_labels = [calendar.month_abbr[i] for i in heat_data.index]
ax3.set_yticklabels(month_labels, rotation=0)

# Titles and labels
ax3.set_title("Average Revenue by Month & Hotel", fontsize=11, fontweight="bold", color="#003580")
ax3.set_xlabel("Hotel Type")
ax3.set_ylabel("Month")

# =============================================================
# (E) Market Segment Share
# =============================================================
ax4 = plt.subplot2grid((3, 4), (2, 2), colspan=2)
segment = df['market_segment'].value_counts(normalize=True).sort_values(ascending=False) * 100
sns.barplot(x=segment.values, y=segment.index, ax=ax4, palette=blue_palette)
ax4.set_title("Market Segment Share (%)", fontsize=11, fontweight="bold", color="#003580")
ax4.set_xlabel("Share (%)")
ax4.set_ylabel("")
ax4.grid(axis='x', alpha=0.3)

plt.tight_layout(pad=3.8)
plt.show()


````
