# **Project Summary -**

In today’s competitive e-commerce landscape, understanding customer behavior is essential for businesses to improve customer satisfaction, increase revenue, and design targeted marketing strategies. Myntra Gifts Ltd., a UK-based division of Myntra specializing in online giftware retail, generates a vast amount of transactional data through its non-store online platform. This project focuses on analyzing that data to uncover meaningful insights and segment customers into distinct groups using machine learning techniques.

The dataset used in this project contains transactional records from December 2009 to December 2011, including details such as invoice number, product description, quantity purchased, unit price, customer ID, and country. Since the dataset does not contain a predefined target variable, the problem is approached as an unsupervised learning task, specifically customer segmentation using clustering techniques.

The project begins with data preprocessing, which involves handling missing values, removing duplicate records, and filtering out invalid transactions such as negative quantities that represent product returns. Data type conversions are also performed to ensure proper handling of date and numerical features. This step is critical to ensure data quality and reliability before further analysis.

Feature engineering plays a central role in this project through the implementation of RFM (Recency, Frequency, Monetary) analysis. Recency measures how recently a customer made a purchase, Frequency indicates how often they purchase, and Monetary represents the total amount spent by the customer. These features provide a comprehensive view of customer behavior and are widely used in marketing analytics for segmentation and targeting.

Exploratory Data Analysis (EDA) is conducted to understand purchasing patterns, product performance, and customer distribution. Visualizations such as histograms, bar charts, and time-series plots are used to identify trends, seasonal patterns, and high-value customer groups. This step helps in gaining insights into customer behavior and guides further modeling decisions.

To prepare the data for clustering, the RFM features are scaled using standardization techniques to ensure all variables contribute equally to the model. The K-Means clustering algorithm is then applied to segment customers into distinct groups based on their purchasing behavior. The optimal number of clusters is determined using the Elbow Method and validated using the Silhouette Score, ensuring a balance between model performance and interpretability.

Once clustering is complete, each segment is analyzed to understand its characteristics. Customers are categorized into groups such as high-value customers, frequent buyers, low-value customers, and at-risk customers. These segments provide actionable insights that can help businesses tailor their strategies effectively.

Based on the analysis, several business recommendations are proposed. High-value customers can be targeted with loyalty programs and exclusive offers, while at-risk customers can be re-engaged through discounts and personalized marketing campaigns. Frequent buyers can be encouraged through upselling and cross-selling strategies, and low-value customers can be nurtured to increase engagement.

In conclusion, this project demonstrates the power of data-driven decision-making in e-commerce. By leveraging machine learning and customer segmentation techniques, businesses can gain a deeper understanding of their customers and implement targeted strategies to enhance customer retention, optimize inventory management, and drive overall growth.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ===============================
# 2.1 CREATE BACKUP COPY
# ===============================

# Always create a backup before modifying the dataset
df_clean = df.copy()

print("Backup created successfully")
````

````python
# ===============================
# 2.2 HANDLE MISSING VALUES
# ===============================

# Check missing values
print("Missing Values Before Cleaning:\n", df_clean.isnull().sum())

# Drop rows where CustomerID is missing
# Reason: CustomerID is essential for segmentation
df_clean = df_clean.dropna(subset=['CustomerID'])

# Drop rows where Description is missing
df_clean = df_clean.dropna(subset=['Description'])

print("\nMissing Values After Cleaning:\n", df_clean.isnull().sum())
````

````python
# ===============================
# 2.3 REMOVE DUPLICATES
# ===============================

# Check duplicates
duplicate_count = df_clean.duplicated().sum()
print("Duplicate Rows Before:", duplicate_count)

# Remove duplicates
df_clean = df_clean.drop_duplicates()

# Verify
duplicate_count_after = df_clean.duplicated().sum()
print("Duplicate Rows After:", duplicate_count_after)
````

````python
# ===============================
# 2.4 DATA TYPE CONVERSION
# ===============================

try:
    # Convert InvoiceDate to datetime
    df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

    print(" InvoiceDate converted successfully")

except Exception as e:
    print("Error in date conversion:", e)
````

````python
# ===============================
# 2.5 REMOVE NEGATIVE VALUES
# ===============================

# Count negative Quantity
negative_qty = df_clean[df_clean['Quantity'] < 0].shape[0]
print("Negative Quantity Rows:", negative_qty)

# Remove negative Quantity (returns/cancellations)
df_clean = df_clean[df_clean['Quantity'] > 0]

# Remove negative UnitPrice (invalid data)
df_clean = df_clean[df_clean['UnitPrice'] > 0]

print("Negative values removed")
````

````python
# ===============================
# 2.6 FEATURE ENGINEERING
# ===============================

# Create TotalPrice column
df_clean['TotalPrice'] = df_clean['Quantity'] * df_clean['UnitPrice']

print("TotalPrice feature created")
````

````python
# ===============================
# CREATE RFM FEATURES
# ===============================

# Ensure InvoiceDate is datetime
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

# Create snapshot date (latest date + 1)
snapshot_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)

# Create RFM table
rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',                                   # Frequency
    'TotalPrice': 'sum'                                       # Monetary
})

# Rename columns
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# Preview
rfm.head()
````

````python
# ===============================
# 2.7 OUTLIER HANDLING (IQR)
# ===============================

# Function to remove outliers using IQR
def remove_outliers(df, column):

    Q1 = df[column].quantile(0.25)   # First quartile
    Q3 = df[column].quantile(0.75)   # Third quartile
    IQR = Q3 - Q1                    # Interquartile range

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter data within bounds
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Apply outlier removal on key columns
for col in ['Quantity', 'UnitPrice', 'TotalPrice']:
    df_clean = remove_outliers(df_clean, col)

print("Outliers removed using IQR method")
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

There is a significant relationship between Quantity and Total Price.
1. Hypothesis
Null Hypothesis (H₀):
There is no significant relationship between Quantity and Total Price.
Alternate Hypothesis (H₁):
There is a significant relationship between Quantity and Total Price.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr

# Hypothesis 1: Quantity vs TotalPrice
corr1, p_value1 = pearsonr(df_clean['Quantity'], df_clean['TotalPrice'])

print("Hypothesis 1: Quantity vs TotalPrice")
print("Correlation:", corr1)
print("P-value:", p_value1)

if p_value1 < 0.05:
    print("Reject Null Hypothesis (Significant Relationship)")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

I used the Pearson Correlation Test (scipy.stats.pearsonr) to obtain the p-value.

##### Why did you choose the specific statistical test?

The Pearson Correlation Test was used to calculate the p-value because both variables are continuous and the objective was to measure the strength and significance of their linear relationship. The test confirmed a statistically significant positive relationship between Quantity and Total Price.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Unit Price has no effect on Quantity purchased.
1. Hypothesis
Null Hypothesis (H₀):
Unit Price has no significant effect on Quantity.
Alternate Hypothesis (H₁):
Unit Price has a significant effect on Quantity.

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 2: UnitPrice vs Quantity
corr2, p_value2 = pearsonr(df_clean['UnitPrice'], df_clean['Quantity'])

print("\nHypothesis 2: UnitPrice vs Quantity")
print("Correlation:", corr2)
print("P-value:", p_value2)

if p_value2 < 0.05:
    print("Reject Null Hypothesis (Significant Relationship)")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

I used the Pearson Correlation Test (scipy.stats.pearsonr) to calculate the p-value.

##### Why did you choose the specific statistical test?

The Pearson Correlation Test was used because both Unit Price and Quantity are continuous variables, and the objective was to measure the strength and significance of their relationship. The test revealed a statistically significant negative correlation, indicating that as Unit Price increases, Quantity decreases, confirming price-sensitive customer behavior.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Average Total Price differs significantly across countries.
Null Hypothesis (H₀):
The mean Total Price is the same across all countries.
Alternate Hypothesis (H₁):
The mean Total Price differs across countries.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import f_oneway

# Group data by country
countries = df_clean['Country'].unique()

grouped_data = [
    df_clean[df_clean['Country'] == country]['TotalPrice']
    for country in countries
]

# Perform ANOVA
f_stat, p_value3 = f_oneway(*grouped_data)

print("\nHypothesis 3: Country vs TotalPrice")
print("F-statistic:", f_stat)
print("P-value:", p_value3)

if p_value3 < 0.05:
    print("Reject Null Hypothesis (Significant Difference)")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

I used the One-Way ANOVA test (scipy.stats.f_oneway) to calculate the p-value.

##### Why did you choose the specific statistical test?

The One-Way ANOVA test was used because the goal was to compare the mean Total Price across multiple countries. Since Country is a categorical variable with more than two groups and Total Price is numerical, ANOVA is the most appropriate test. The results show a statistically significant difference in average Total Price across countries, leading to rejection of the null hypothesis.

#### 1. Elbow Method

````python
# ===============================
# ELBOW METHOD
# ===============================

wcss = []

for i in range(2,10):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(rfm_scaled)
    wcss.append(model.inertia_)

# Plot
plt.figure(figsize=(8,5))
plt.plot(range(2,10), wcss, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
````

🔹 Why did you use this chart?

The Elbow Method is used to determine the optimal number of clusters (K) by analyzing the Within-Cluster Sum of Squares (WCSS). It helps identify the point where adding more clusters does not significantly improve model performance.

🔹 Insights from the Chart
WCSS decreases as the number of clusters increases
A sharp drop is observed from K = 2 to K = 4
After K = 4 or 5, the curve starts to flatten

👉 Key Observation:

The “elbow point” appears around K = 4

🔹 Business Interpretation
Up to K = 4 → meaningful segmentation
Beyond that → diminishing returns

👉 Too many clusters:

Difficult to interpret
Hard to apply business strategies

**Silhouette Score**

````python
# ===============================
# SILHOUETTE SCORE
# ===============================

from sklearn.metrics import silhouette_score

for i in range(2,10):
    model = KMeans(n_clusters=i, random_state=42)
    labels = model.fit_predict(rfm_scaled)
    score = silhouette_score(rfm_scaled, labels)
    print(f"Clusters: {i}, Silhouette Score: {score}")
````

Hyperparameter Tuning

````python
# ===============================
# HYPERPARAMETER TUNING (K-MEANS)
# ===============================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

best_score = -1
best_params = {}

for k in range(2,10):
    for init in ['k-means++', 'random']:

        model = KMeans(
            n_clusters=k,
            init=init,
            n_init=10,
            max_iter=300,
            random_state=42
        )

        labels = model.fit_predict(rfm_scaled)

        score = silhouette_score(rfm_scaled, labels)

        if score > best_score:
            best_score = score
            best_params = {'n_clusters': k, 'init': init}

print("Best Parameters:", best_params)
print("Best Silhouette Score:", best_score)
````

Final Optimized Model

````python
# ===============================
# FINAL MODEL WITH BEST PARAMS
# ===============================

kmeans_final = KMeans(
    n_clusters=best_params['n_clusters'],
    init=best_params['init'],
    n_init=10,
    max_iter=300,
    random_state=42
)

rfm['Cluster'] = kmeans_final.fit_predict(rfm_scaled)
````

📌 Which hyperparameter optimization technique have you used and why?


Hyperparameter tuning was performed using a manual grid search approach by testing different values of the number of clusters (K) and initialization methods. The Silhouette Score was used as the evaluation metric to select the optimal parameters. This approach is suitable for clustering problems where traditional cross-validation cannot be applied.

📌 Have you seen any improvement?


Yes, after hyperparameter tuning, the Silhouette Score improved, indicating better cluster separation and compactness. The optimized parameters resulted in more meaningful and well-defined customer segments compared to the initial model.

**Evaluation Metric Score Chart (Before vs After)**

````python
# Example comparison (you can adjust values)
labels_initial = KMeans(n_clusters=4, random_state=42).fit_predict(rfm_scaled)
score_initial = silhouette_score(rfm_scaled, labels_initial)

labels_tuned = kmeans_final.fit_predict(rfm_scaled)
score_tuned = silhouette_score(rfm_scaled, labels_tuned)

print("Before Tuning:", score_initial)
print("After Tuning:", score_tuned)
````
